# pandas 基本操作-根据索引获取数据

```python
# 导包
import pandas as pd
import os   
```

```python
# 读取文件数据
current_dir = os.getcwd()
os.chdir("E:\python\project\python\pandas")
print(current_dir)
sd=pd.read_csv('./data/stock_day.csv')
sd.head(10)
# print(sourceData)
```

```python
# 根据索引获取具体数据(例如:第一行第一列) [先列后行]
value1=sd["open"]["2018-02-27"]
value2=sd["high"]['2018-02-07']
value2
```

```python
#通过loc属性来获取，只能通过：行索引+列名的方式来获取
# 格式 loc[行索引][列名]或者是loc[行索引,列名]
cr1=sd.loc["2018-02-27"]['close']
cr2=sd.loc["2018-02-27","low"]
cr2
```

```python
# 需求：获取从2018-02-27 - 2018-02-22的open列数据
# 格式  ====> 数据源.loc[起始行索引:结束行索引]
range_row=sd.loc["2018-02-27":"2018-02-22"]
# 需求：获取从2018-02-27 - 2018-02-22的open, high, close, low列数据
# 格式 ====>数据源.loc[起始行索引:结束行索引,列名(适用于单列情况)]  或者是 数据源.loc[起始行索引:结束行索引,[列名1,列名2,列名3.....](适用于多列情况)]
range_row2=sd.loc["2018-02-27":"2018-02-22",["open","high","close","low"]]
range_row2

# 需求：获取从2018-02-27 - 2018-02-22的所有列数据
# 格式====> 数据源.loc[起始行索引:结束行索引] 或者是 数据源.loc[起始行索引:结束行索引, :]  ***** ":" 指起始至结束
range_row3=sd.loc["2018-02-27":"2018-02-22"]
range_row4=sd.loc["2018-02-27":"2018-02-22",:]
range_row4
```
