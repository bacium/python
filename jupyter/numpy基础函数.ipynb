```python
# 导包
import numpy as np
```

```python
# 创建数组
arr=np.random.randn(3,4)
arr
```

```python
#常见内置函数
print(np.ceil(arr)) #向上取整
print(np.floor(arr)) #向下取整
print(np.rint(arr)) # 四舍五入
print(np.multiply(arr,arr)) #矩阵乘法运算(自身相乘，每一项对应相乘)
print(np.divide(arr,arr)) # 矩阵除法运算(自身相除，对应每项相除)
print(np.isnan(arr)) #判断是否为空，True为空，False不为空
print(np.where(arr>0,arr,0))  #numpy的三元运算
```
