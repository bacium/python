## 需求:某地十一时到十二时一小时内每分钟的温度变化

```python
# 数据准备
x=range(60)
y=[random.uniform(3,20) for i in x]

# 创建画布
plt.figure(figsize=(40,30),dpi=80)
plt.plot(x,y)
x_lable=[f"11点{i}分" for i in x]
y_lable=[i for i in range(0,25)]
plt.xticks(x[::5],x_lable[::5])
plt.yticks(y_lable[::5])
plt.rcParams['xtick.labelsize'] = 20  # x轴刻度标签大小
plt.rcParams['ytick.labelsize'] = 20  # y轴刻度标签大小
# 数据补充
# 网格
plt.grid(True,linestyle="dashed",alpha=0.5)
# 标题
plt.title("北京11时至12时每分钟温度变化图",fontsize=30)
# x轴标题或者数据指示含义
plt.xlabel("时间",fontsize=30)
# y轴标题或者数据含义
plt.ylabel("温度",fontsize=30)
# 保存图片
plt.savefig("./images/temprature.png")

plt.show()
```
