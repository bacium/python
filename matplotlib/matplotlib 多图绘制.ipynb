```python
import matplotlib.pyplot as plt
import os
import random
# 解决中文乱码问题.  思路1: 本机(或者虚拟机)安装SimHei字体.    思路2: 导包, 配置字体.
# 设置显示中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']
# 设置正常显示符号
plt.rcParams['axes.unicode_minus'] = False
print(os.getcwd())
os.chdir("E:\python\project\python\matplotlib")
```

```python
# 创建数据
x=[i for i in range(60)]
y_bj=[random.uniform(5,18) for i in range(60)]
print(y_bj)
y_js=[random.uniform(16,23) for i in range(60)]
print(y_js)
```

```python

# 创建画布
plt.figure(figsize=(30,15),dpi=80)
# 渲染图
plt.plot(x,y_bj,color="r",label="北京")
plt.plot(x,y_js,color="b",label="江苏")
# 设置 x,y 坐标轴刻度
x_ticks_label=[f"11时{i}分" for i in range(60)]
y_ticks=[i for i in range(0,30)]
plt.xticks(x[::5],x_ticks_label[::5])
plt.yticks(y_ticks[::5])
# 设置坐标轴文本信息
plt.ylabel("温度",fontsize=15)
plt.xlabel("时间",fontsize=15)

# 图标题
plt.title("北京/江苏温度统计图",fontsize=20)

# 图例
plt.legend(loc=0,fontsize=15)
# 保存图片
plt.savefig('./images/doubleTemp.jpeg')

# 展示图形
plt.show()
```

## 面向对象方式绘图 -> 1个画板多个坐标系

```python
# 创建数据
# 创建数据
x=range(60)
y_bj=[random.uniform(5,20) for i in x]
print(y_bj)
y_js=[random.uniform(12,26) for i in x]
print(y_js)
```

```python
# 创建画布

fig,ax=plt.subplots(figsize=(30,15),nrows=1,ncols=2)
y_ticks = [i for i in range(0, 30)]
# 渲染图
ax[0].plot(x,y_bj,color="r",label="北京")
ax[1].plot(x,y_js,color="b",label="江苏")

# 设置纵坐标名称
ax[0].set_ylabel("温度",fontsize=20)
ax[1].set_ylabel("温度",fontsize=20)
# 设置横坐标名称
ax[0].set_xlabel("时间",fontsize=20)
ax[1].set_xlabel("时间",fontsize=20)

# 设置横坐标刻度
ax[0].set_xticks(x[::5],[f"12时{i}分" for i in x][::5])
ax[1].set_xticks(x[::5],[f"12时{i}分" for i in x][::5])

#设置纵坐标刻度
ax[0].set_yticks(y_ticks[::5])
ax[1].set_yticks(y_ticks[::5])

# 设置网格
ax[0].grid(linestyle="--",alpha=0.5,color="g")
ax[1].grid(linestyle="-.",alpha=0.5,color="g")

# 设置图例
ax[0].legend(loc=0,fontsize=15)
ax[1].legend(loc=0,fontsize=15)

# 设置图形标题
ax[0].set_title("北京中午小时温度",fontsize=25)
ax[1].set_title("江苏中午小时温度",fontsize=25)

# 保存图表
plt.savefig('./images/beijing_jiangsu_temp.jpeg')

plt.show()
```
