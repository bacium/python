```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import  KNeighborsClassifier
```

```python
iris=load_iris()
iris
```

```python
iris_data=pd.DataFrame(iris.data,columns=iris.feature_names)
x=iris.data
y=iris.target
print(f"数据集：{x}")
print(f"标签:{y}")
```

```python
# 数据分割及预处理
x_train,x_test,y_train,y_test=train_test_split(x, y, test_size=0.3,random_state=22)

# 特征工程
transfer=StandardScaler()
x_train=transfer.fit_transform(x_train)
x_test=transfer.transform(x_test)
```

```python
# 模型预测
model=KNeighborsClassifier(n_neighbors=5)
model.fit(x_train,y_train)
y_pre=model.predict(x_test)
y_pre


```

```python
# 模型评估
print(f"模型评估：{accuracy_score(y_test, y_test)}")
print(classification_report(y_test, y_pre, target_names=iris.target_names))


```

```python
# 测试代码
sample = [[5.8, 3.2, 4.9, 1.8]]  # 花萼&花瓣尺寸
sample = transfer.transform(sample)
pred = model.predict(sample)
print("预测结果:", iris.target_names[pred][0])
```
