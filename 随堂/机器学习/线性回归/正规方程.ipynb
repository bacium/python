```python
# 导包
from skimage.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,root_mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd
import os
```

```python
os.getcwd()
os.chdir("E:\python\project\python\随堂\机器学习\线性回归")
```

```python
df=pd.read_csv('./data/house_data.csv')
y_train=df.values[:,-1]
y_train
```

```python
# 数据预处理
x_train,x_test,y_train,y_test=train_test_split(df,y_train, test_size=0.2)
print(f"训练集特征：{x_train}")
print(f"训练集特征：{x_test}")
print(f"测试集特征：{y_train}")
print(f"测试集标签：{y_test}")


# 特征工程
transfer=StandardScaler()
x_train=transfer.fit_transform(x_train)
x_test=transfer.transform(x_test)



```

```python
# 模型训练
model=LinearRegression()
model.fit(x_train,y_train)
y_pre=model.predict(x_test)
print(f"预测结果：{y_pre}")
print(f"真实结果：{y_test}")
print(f"预测值与真实值的差异：{y_pre - y_test}")
print(f"模型准确度:{model.coef_}")
print(f"模型偏置:{model.intercept_}")
```

```python


# 模型评估
print(f"均方误差：{mean_squared_error(y_test,y_pre)}")
print(f"平均绝对误差：{mean_absolute_error(y_test, y_pre)}")
print(f"均方根误差：{root_mean_squared_error(y_test,y_pre)}")



```
