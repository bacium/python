```python
import torch
```

```python
# 张量的加减乘除运算
"""
    加法:add和add_
    减法:sub和sub_
    乘法:mul和mul_
    除法:dev和div_

"""
torch.manual_seed(20251211)
t = torch.randint(0, 10, (2, 3))

# 加法
"""
    每个元素都和对应数据相加
"""
t1 = torch.add(t, 10)
# print(t1)
# 减法
t2 = torch.sub(t, 1)

# 乘法
t3 = torch.mul(t, 2)
t3_ = torch.randint(0, 8, (2, 3))
print(f"t3_:{t3_}")
mt3 = t3 * t3_
print(f"mt3的积:{mt3}")

# 除法
t4 = torch.div(t, 2)
t4


```

```python
# 矩阵的乘法运算
# 矩阵A和B相乘 A行B列,行和列相乘
t1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
t2 = torch.tensor([[1, 2], [3, 4], [5, 6]])

t3 = t1 @ t2
print(t3)
t4= torch.matmul(t1, t2)
t4 # 结果同t3
```
