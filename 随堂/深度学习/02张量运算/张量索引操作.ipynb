```python
import torch

torch.manual_seed(666)
```

```python
# 行列索引的使用
# 索引表达式:张量[行,列]
t1 = torch.randint(0, 10, (3, 4))
print(t1)
print(t1[0, 1])  #  取第一行第二列的值
print(t1[1, :])  # 取第二行所有值
print(t1[:, 1])  # 取第二列所有制
```

```python
t1 = torch.randint(0, 10, (2, 3, 4))
print(t1)  # 张量维度:2个3行四列
t1[0, 1, 2]  # 取第一块,第二行第三列的值
```
