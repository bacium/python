# 创建线性张量

```python
import torch
import numpy as np
```

```python
# torch.arange(start, end, step)
print(torch.arange(1,10))
print(torch.arange(1,10,2))
```

```python
# torch.linspace(start, end, steps)
print(torch.linspace(1,10,5))  # 等差数列张量
print(torch.linspace(1,10,5,dtype=torch.int64))
```

```python
# 设置和获取随机种子
# 设置随机种子
torch.random.manual_seed(100)

print(torch.rand(2, 3))
print(torch.rand(2, 3))
"""
    执行结果:
    tensor([[0.1117, 0.8158, 0.2626],
        [0.4839, 0.6765, 0.7539]])
    tensor([[0.2627, 0.0428, 0.2080],
            [0.1180, 0.1217, 0.7356]])
    结论:设置了随机种子之后,无论运行上述代码多少次,结果都是不变的
"""

# 获取设置的随机种子

print(torch.random.initial_seed()) # 获取100上面设置的随机种子
```

```python
# 随机张量
print(torch.rand(2, 3)) # 随机两行三列的浮点数张量
print(torch.randn(2, 3)) # 随机两行三列的浮点数张量

"""
     randint随机两行三列的整型张量
     参数1: 最小值
     参数2:最大值(取不到)
     参数3:形状

"""
print(torch.randint(5, 10, (2, 3))) 
 
```
