张量和numpy的转换

```python
import numpy as np
import torch
```

```python
"""
    numpy转化为张量:
    方法一:张量.from_numpy() ====>t0和t1的值除了第一个值不同其他值相同,即修改numpy的值,张量的值也随之修改,即numpy和张量共享内存地址
    方法二:torch,tensor() =====> t2的值即使修改了numpy数据n的第一个值,张量的值依然时不变的,即:通过tensor(data_numpy)转化的张量占用单独的内存,不共享地址.
"""
a = np.random.randn(4)
print(a) 
t0=torch.from_numpy(a)
print(t0)

# 修改第一个值
a[0]=100
t1= torch.from_numpy(a)
print(t1) # 其他值不变,第一个值为100,
"""
    使用tensor()方法转化
    t2的值即使修改了numpy数据n的第一个值,张量的值依然时不变的,即:通过tensor(data_numpy)转化的张量占用单独的内存,不共享地址.
"""
n=np.random.randn(4)
n[0]=100
t2=torch.tensor(n)
print(t2) # t2的值即使修改了numpy数据n的第一个值,张量的值依然时不变的,
```

```python
"""
    张量转换为numpy
    方式一:张量.numpy()
    方式二:张量.numpy().copy()

"""
t2=torch.tensor([1,2,3])
t3=t2.numpy()
print(t3) # [1,2,3]
t5=t2.numpy().copy()
t2[0]=100
t4=t2.numpy()
print(t4) # [100,2,3] ====>此时张量t2的值已经修改了,numpy的值也跟着修改了.即共享内存地址.
print(t5) # [1,2,3] ====> numpy的值不会随张量的值改变而改变.即不共享内存地址.
```
