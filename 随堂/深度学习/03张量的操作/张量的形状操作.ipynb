# 张量的形状操作

```python
import torch
torch.manual_seed(100)
```

```python
t1=torch.tensor([[1,2,3],[4,5,6]])
print(t1.shape)
t1
```

```python
# 改变张量的形状
t2=t1.reshape(3,2)
t3=t1.reshape(-1)
t3
```

```python
t4=t1.unsqueeze(dim=0)
print(t4)
# t4 在0维增加一维 
[
  [
    [1, 2, 3],
    [4, 5, 6]
  ]
]
t4.size()
```

```python
t5=t1.unsqueeze(dim=1)
print(t5)

# t5 列增加一个维度
[
  [
    [1, 2, 3]
  ],

  [
    [4, 5, 6]
  ]
]

t5.size()
```

```python
t6=torch.randint(0,10,(2,1,3,1))
print(t6)
t6.size()
```

```python
# squeeze删除所有维度为1的维
t7=t6.squeeze()
t7
```
