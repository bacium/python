# 张量创建

```python
import numpy as np
import torch
```

```python
print(torch.tensor(10))
```

```python
print(torch.tensor([1,2,3]))
```

```python
arr=np.array([[1,2,3],[4,5,6]])
print(torch.tensor(arr))
```

```python
print(torch.Tensor(10))
```

```python
print(torch.Tensor([1,2,3]))
```

```python
print(torch.Tensor(arr))
```
