# dataFrame 常用属性及方法

## 1 常用属性

```python
# 导包
import pandas as pd
import numpy as np
```

```python
arr=np.random.randint(40,100,size=(10,5),dtype=np.int32)
df=pd.DataFrame(arr,columns=["语文","数学","英语","生物","化学"],index=["同学"+ str(i) for i in range(10)])
```

```python
# 常见属性
# 属性1: index 获取数据的行索引
df.index
```

```python
# 属性2: columns 获取数据的列名
df.columns
```

```python
# 属性3: values 获取数据的所有值
df.values
```

```python
# 属性4 size 获取数据的长度
df.size
```

```python
# 属性5 dtypes 数据元素的类型(取决于定义时的数据类型,此处由numpy定义)
df.dtypes
```
