# dataFrame 常用属性及方法

## 1 常用属性

```python
# 导包
import pandas as pd
import numpy as np
```

```python
arr=np.random.randint(40,100,size=(10,5),dtype=np.int32)
df=pd.DataFrame(arr,columns=["语文","数学","英语","生物","化学"],index=["同学"+ str(i) for i in range(10)])
df
```

```python
# 常见属性
# 属性1: index 获取数据的行索引
df.index
```

```python
# 属性2: columns 获取数据的列名
df.columns
```

```python
# 属性3: values 获取数据的所有值
df.values
```

```python
# 属性4 size 获取数据的长度
df.size
```

```python
# 属性5 dtypes 数据元素的类型(取决于定义时的数据类型,此处由numpy定义)
df.dtypes
```

```python
# 属性6 T 数据转置 (将数据的行和列互换,数据也随之互换)
df.T
```

## 2 常用方法

```python
# 方法1 head() 查看数据前()行数据,默认5行
df.head()
```

```python

# 方法2 tail() 查看数据后()行数据,默认5行
df.tail(10)
```

```python
# 方法3 describe() 获取数据描述信息
df.describe()
```

```python
# 方法4 info() 获取数据详细信息
df.info()
```

## 4 DataFrame 的数据类型


```python
#datetime 时间日期类型
date=pd.to_datetime("2018-01-01")
date
```

```python
# timedelta 时间差类型
start_time=pd.to_datetime("2018-01-10")
end_time=pd.to_datetime("2018-01-20")
print(end_time-start_time)
print(type(end_time-start_time))
```

```python
# category 类型 ,只保存索引和唯一值
fruit=pd.Series(["apple","banana","cherry","apple","banana","peach","orange"],dtype="category")
print(fruit)
print(type(fruit))
```
