# pandas常用数据类型及方法


## 1.1 Series的创建

```python
# 1 导包
import pandas as pd
import numpy as np
```

```python
# 通过列表创建Series对象
sd=pd.Series([1,2,3,4])
print(sd)
print(type(sd)) #'pandas.core.series.Series'
```

```python
#方式2: 把Python列表 -> Series对象, 指定索引.
sd1=pd.Series([1,2,3,4,5,6,7,8,9],index=['a','b','c','d','e','f','g','h',"i"])
print(sd1)
```

```python
#方式3: 把Python元组 -> Series对象.
sd2=pd.Series(("李四","男","北京"),index=["name","gender","address"])
print(sd2)
```

```python
# 方式4: 把字典 -> Series对象. 键: 索引, 值: 数据.
sd3=pd.Series({"name":"张三","age":23,"address":"北京"})
print(sd3)
```

```python
#方式5: 把Numpy的ndarray对象 -> Series对象.
arr=np.array(["a","b","c","d","e","f","g","h","i"])
sd4=pd.Series(arr)
print(sd4)
```

## 1.2 Series的常见属性

```python
# 创建一个Series对象
sd5=pd.Series(["张三","李四","王五","赵六","吴麻子"],index=['a','b','c','d','e'])
print(sd5)
```

```python

# 属性1 获取索引index
print(sd5.index) #Index(['a', 'b', 'c', 'd', 'e'], dtype='object') 
```

```python
# 属性2  获取Series对象的值
print(sd5.values) #['张三' '李四' '王五' '赵六' '吴麻子']
```

```python
# 属性3 根据索引获取Series对象的值
print(sd5["c"])
```
