## 绘制多国曲线

```python
china_df=data[data.country=="中国"]
usa_df=data[data.country=="美国"]
jp_df=data[data.country=="日本"]
print(china_df.head(10))
print(usa_df.head(10))
print(jp_df.head(10))
```

```python
# 绘制曲线 三条曲线
china_df=data[data.country=="中国"]
usa_df=data[data.country=="美国"]
jp_df=data[data.country=="日本"]
china_df.plot()
usa_df.plot()
jp_df.plot()
```

```python
# 绘制到一张图中
china_df.GDP.plot()
usa_df.GDP.plot()
jp_df.GDP.plot()
```

```python
# 追加图例
china_df=data[data.country=="中国"].copy()
usa_df=data[data.country=="美国"].copy()
jp_df=data[data.country=="日本"].copy()

china_df.rename(columns={"GDP":"China"}, inplace=True)
usa_df.rename(columns={"GDP":"Usa"}, inplace=True)
jp_df.rename(columns={"GDP":"Jp"}, inplace=True)
print(china_df.head(10))
print(usa_df.head(10))
print(jp_df.head(10))
```

```python
# 设置图例 英文
china_df.China.plot(color="red",legend=True)
jp_df.Jp.plot(color="yellow",legend=True)
usa_df.Usa.plot(color="blue",legend=True)
```

```python
# 设置图例和颜色，中文
china_df=data[data.country=="中国"].copy()
usa_df=data[data.country=="美国"].copy()
jp_df=data[data.country=="日本"].copy()
china_df.rename(columns={"GDP":"中国"}, inplace=True)
usa_df.rename(columns={"GDP":"美国"}, inplace=True)
jp_df.rename(columns={"GDP":"日本"}, inplace=True)
china_df.中国.plot(color="red",legend=True)
usa_df.美国.plot(color="blue",legend=True)
jp_df.日本.plot(color="yellow",legend=True)
```
