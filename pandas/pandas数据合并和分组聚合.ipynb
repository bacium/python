# 分组聚合

```python
# 获取数据
ds=pd.read_csv('./data/uniqlo.csv')
ds
```

```python
# 使用groupby函数进行分组,得到的数据类型为DataFrameGroupBy 类型
# 单列分组
by_city=ds.groupby("city")
# 通过获取分组的信息来查看具体数据
by_city.get_group("北京").head() # 获取到所有属于北京的数据
```

```python
# 基于多列分组
multiple_group=ds.groupby(["city","channel"])
multiple_group.get_group(('北京',"线下")).head(20)
# 获取第一条或者最后一条数据
multiple_group.first()
multiple_group.last()
```

```python
# 格式: ds.groupby(['分组字段1', '字段2'...]).agg({'列名1':'聚合函数名', '列名2':'聚合函数名'...})
ds.groupby(["revenue","order"]).agg({"customer":"sum",'unit_cost':"mean"})
```
