```python
# 导包
import pandas as pd
import os
os.chdir('E:\python\project\python\pandas')
ds=pd.read_csv('./data/stock_day.csv')
ds.head(10)
```

```python
## 逻辑运算符号
ds.head(5).loc[["2018-02-27","2018-02-23","2018-02-22"]]
ds.head(5).iloc[[0,2,3]]
ds.head()[[True,False,True,True,False]]
```

```python
# 需求1: 筛选出ds中, 所有open价格在23以上的数据
ds[ds["open"]>=23]
```

```python
# 需求2: 筛选出ds中, 所有open价格在[23, 24]之间的数据
res=ds[(ds["open"]>=23) & (ds['open']<=24)]

#验证:排序一下看看取得数据是否和结果能对应
res.sort_values("open",ascending=True)
```

## 逻辑运算函数 

```python
# query(直接写比较表达式即可.)
# 需求: 筛选出df中, 所有open价格在[23, 24]之间的数据
ds.query("open>=23 & open<=24")
```

```python
# 3. isin([值1, 值2....]) 判断是否是其中的任意1个值
ds
# 需求: 筛选出df中价格为 23.80 或者22.61的数据.
ds.query("open==23.80 or open==22.61")
ds[(ds["open"]==23.80) | (ds["open"]==22.61)]
ds.query("open in [23.80,22.61]")

# 使用isin函数
ds[ds["open"].isin([23.80,22.61])]
```
