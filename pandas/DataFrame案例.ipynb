# DataFrame 学习案例

```python
#1 引入pandas 
import pandas as pd
import numpy as np
```

```python
# 方式1: 把字典列表 -> DataFrame对象.   每组键值对 = 1列数据.
# 1. 定义数据集, 字典 + 列表的方式.
dict_list={
    "日期":["2025-10-11","2025-10-12","2025-10-13","2025-10-14"],
    "温度":["36","32","37","31"],
    "湿度":[65,64,56,63]
}
df=pd.DataFrame(dict_list,index=["a","b","c","d"])
print(df)
```

```python
# 方式2: 把 列表 + 元组 => DataFrame对象.    列表的每个元素(元组) = 一行数据.
# 1. 定义数据集, 列表 + 元组的方式.
list_tump=[
    ("2025-10-11","36",65),
    ("2025-10-12","34",60),
    ("2025-10-13","37",56),
]
df1=pd.DataFrame(list_tump,columns=["时间","温度","湿度"],index=['a','b','c'])
print(df1)
```

```python
# 方式3: 把 Numpy的ndarray对象 -> DataFrame对象.
arr=np.random.uniform(-1,7,size=9).reshape(3,3)
df3=pd.DataFrame(arr,columns=["分数"+i for i in "123"],index=[ i for i in "ABC"])
print(df3)
```
