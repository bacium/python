# pandas缺失值处理

```python
import pandas as pd
import os
import numpy as np
os.getcwd()
os.chdir("E:\python\project\python\pandas")
```

```python
ds=pd.read_csv('./data/movie.csv')
ds
```

```python
# 查看对象信息
# ds.info() # info信息中有部分值不是1000个,也就是指对应的数据中含有确实的空值情况
# ds.isnull() # 判断每列的数据是否为空
# ds.notnull() # 判断每列的数据是否为非空数据
ds.isnull().sum()  # 统计为空字段的总和
ds.notnull().sum() # 统计非空字段的总和
```

## 删除缺失值

```python

ds.dropna(inplace=True,axis=1) # axis=0删除含有缺失值的行 ,axis=1 删除多有含有缺失值的列
ds.isnull().sum()   #所有结果为0,代表缺失值全部被删除
ds.info()
ds
```

```python
## 填充缺失值
ds.fillna(55,axis=1,inplace=True) 
ds.info() # 所有信息都是1000,代表确实值都填充完毕
```

```python
# 3. 用 每列的平均值, 来填充 每列的缺失值.
# 3.1 看看有缺失值的两列的 平均值.
# ds.info()
print(ds["Revenue (Millions)"].mean()) # 82.95637614678898
print(ds["Metascore"].mean()) # 58.98504273504273

for col_name in ds.columns:
    # print(ds[col_name])
    if ds[col_name].isnull().sum()>0:
        ds[col_name].fillna(ds[col_name].mean(),inplace=True)

ds
```
